```python
import matplotlib.pyplot as plt
import numpy as np
#Title and label
ypoints = np.array([3, 8, 1, 10, 5, 7])
print(plt.plot(ypoints, linestyle='dashed'))
plt.title('Sample graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
print(plt.show())
```

```python
#Marker,color,Linewidth
ypoints = np.array([3, 8, 1, 10, 5, 7])
print(plt.plot(ypoints, marker = 'o'))
print(plt.show())

xpoints = np.array([0, 6])
ypoints = np.array([0, 250])

print(plt.plot(xpoints, ypoints, 'violet'))

print(plt.show())
```

```python
#sub_plot
#plot 1
x = np.array([0,1,2,3,4,5])
y = np.array([3,8,1,10,9,4])
plt.subplot(1,2,2)
plt.plot(x,y)
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(["A", "B", "C", "D"])
y = np.array([5, 7, 2, 15])

plt.bar(x,y)
plt.show()
```

```python
plt.bar([0.5,2.5,4.5,6.5,8.5],[50,40,70,80,20],
label="BMW",width=1)
plt.bar([1.5,3.5,5.5,7.5,9.5], [80,20,20,50,60],
label="Audi", color='c', width=1)
plt.legend()
plt.xlabel('Days')
plt.ylabel('Distance (kms)')
plt.title('Information')
plt.show()
```

```python
#Horizontal bar
plt.barh(x,y)
plt.show()
```

```python
#Color option
plt.bar(x,y, color='red')
plt.show()
```

```python
#width option
plt.bar(x,y, color='g', width=0.1)
plt.show()
```

```python
#Histogram
x = np.random.normal(300, 20, 250)
plt.hist(x)
plt.show()
```

```python
#Pie chart
y = np.array([40, 30, 20, 10])
plt.pie(y)
plt.show()
```

```python
#With Labels
y = np.array([35, 25, 25, 15])
mylabels = ["MPMC", "PPML", "ContolSystem", "PSA"]
plt.pie(y, labels = mylabels)
plt.show()
```

```python
#Explode
y = np.array([35, 25, 25, 15])
mylabels = ["MPMC", "PPML", "ContolSystem", "PSA"]

myexplode = [0, 0, 0.3, 0]
plt.pie(y,labels=mylabels,explode=myexplode)
plt.show()
```

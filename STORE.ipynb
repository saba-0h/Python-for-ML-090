```python
import pandas as pd
df=pd.DataFrame([[1,2,3],[4,5,6],[7,8,9]],index=['one','two','three'],columns=['a','b','c'])
print(df)
```

```python
#taking the input from dataframe and storing in the excel file
print("Taking the input from Dataframe and storing in the excel file")
df.to_excel('pandastoexcel.xlsx',sheet_name='first.xlsx')
```

```python
#second dataframe
dff=pd.DataFrame([[100,212,312],[414,502,614],[702,812,923]],index=['four','five','six'],columns=['a','b','c'])
print(dff)
dff.to_excel('pandastoexcel2.xlsx',sheet_name='first.xlsx')
```

```python
x=pd.read_excel('pandastoexcel.xlsx')
y=pd.read_excel('pandastoexcel2.xlsx')
z=pd.concat([x,y])
z.to_excel('excel3.xlsx')
```

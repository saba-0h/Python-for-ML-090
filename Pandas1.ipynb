```python
import pandas as pd
df=pd.DataFrame()
print(df)
```

```python
emp=pd.Series(['Saba','Parama','Ananth'])
id=pd.Series([90,72,123])
frame={'Emp':emp,'ID':id}
result=pd.DataFrame(frame)
print(result)
```

```python
print(result['Emp'])
```

```python
#adding new col.
result['Age']=pd.Series([18,19,18])
print(result)
```

```python
#deleting columns
del result['Age']
print(result)
```

```python
#3rd row exraction
print(result.loc[2])
```

```python
#slcing of rows
print(result[1:3])
```

```python
d2=pd.DataFrame([['Yuvan', 125],['Pranav AP', 77]], columns=['Emp','ID'])
result=pd.concat([result,d2], ignore_index=True)
print(result)
```
